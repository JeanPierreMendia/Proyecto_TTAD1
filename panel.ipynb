```python
import json
from pathlib import Path
```

```python
rutas = [
    Path("data.json"),
    Path("Proyecto_TTAD_1/data.json")
]

ruta = next((p for p in rutas if p.is_file()), None)
with open(ruta, "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]
print(f"Registros de ventas: {len(ventas)}")
```

```python
ingresos = {}

for venta in ventas:
    producto = venta["producto"]
    importe = venta["cantidad"] * venta["precio_unitario"]
    ingresos[producto] = ingresos.get(producto, 0) + importe

maximo = max(ingresos.values())

for producto, importe in ingresos.items():
    if importe == maximo:
        print(f"{producto}: {importe:.2f}")
```

```python
vendedor = "Ana"

ventas_filtradas = [
    venta for venta in ventas
    if venta["vendedor"] == vendedor
]

for venta in ventas_filtradas:
    print(venta["producto"], venta["cantidad"])
```

```python
total = sum(
    venta["cantidad"] * venta["precio_unitario"]
    for venta in ventas
)

ticket_medio = total / len(ventas) if ventas else 0

print(f"Ticket medio: {ticket_medio:.2f}")
```
